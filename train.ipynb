```python
X = df.drop(columns='is_fraud')
y = df['is_fraud']
```

## 1.3: Encoding Categorical and Bool data

```python
X_object = X.select_dtypes(exclude=["number", "bool"])
X_object.head()
```

```python
X_bool = X.select_dtypes(include=['bool'])
X_bool.head()
```

```python
object_cols = X_object.columns
bool_cols = X_bool.columns
X_numerical = X.drop(columns=object_cols.to_list() + bool_cols.to_list())
X_numerical.head()
```

```python
#Encode bool data
def bool_encode(data, cols):
    data = data.copy()
    for col in cols:
        data[col] = data[col].astype(int)
    return data

#Encode categorical data
def categorical_encode(data, cols):
    data = data.copy()
    for col in cols:
        data[col] = pd.factorize(data[col], sort=True)[0]
    return data

# Ham tren dung cho mot DataFrame; train/validation dung chung
# OrdinalEncoder o phan Feature Engineering de giu cung bang ma.
```

```python
X_bool = bool_encode(X_bool, bool_cols)
X_bool.head()
```

## 1.4: Feature Engineering

```python
X_features = pd.concat([X_numerical, X_object, X_bool], axis=1)
X_features.head()
```

```python
y_train = y
y_train
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_features, y.astype(int), test_size=0.2, random_state=42, stratify=y
)
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Hoc ma categorical tren train, dung cung bang ma cho validation
categorical_cols = object_cols.to_list()
category_encoder = OrdinalEncoder(
    handle_unknown='use_encoded_value',
    unknown_value=-1,
    encoded_missing_value=-1,
    dtype='int64',
)

X_train = X_train.copy()
X_test = X_test.copy()
X_train[categorical_cols] = category_encoder.fit_transform(X_train[categorical_cols])
X_test[categorical_cols] = category_encoder.transform(X_test[categorical_cols])

# Xem ma so cua tung gia tri; nhom moi hoac bi thieu duoc gan -1
category_mapping = {
    col: {value: code for code, value in enumerate(categories) if pd.notna(value)}
    for col, categories in zip(categorical_cols, category_encoder.categories_)
}
print(category_mapping)
print(X_train[categorical_cols].dtypes)
X_train.head()
```

# 2: Model

#import libraries

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier

# Cac cot categorical da duoc encode thanh int o Feature Engineering
# Gioi han do sau de giam overfitting
model = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('tree', DecisionTreeClassifier(
        max_depth=6,
        min_samples_leaf=50,
        class_weight='balanced',
        random_state=42,
    )),
])

model.fit(X_train, y_train)
print('Da huan luyen xong Decision Tree!')
```

# 3: Evaluate

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# X_test la tap validation duoc tach tu train.csv
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

print('ROC-AUC:', round(roc_auc_score(y_test, y_proba), 4))
print(classification_report(y_test, y_pred, digits=4, zero_division=0))
print('Confusion matrix (hang: thuc te, cot: du doan; thu tu lop: 0, 1):')
print(confusion_matrix(y_test, y_pred))
```

# 4: Xuất xác suất ra CSV
Dự đoán xác suất `is_fraud = 1` cho `data/test.csv` bằng model đã huấn luyện. Dùng lại encoder của tập train và lưu `submission.csv` theo thứ tự ID trong file mẫu.

```python
test_df = pd.read_csv('data/test.csv')
submission = pd.read_csv('data/sample_submission.csv')

# Giu dung thu tu cot va bang ma da dung khi train
X_submit = test_df[X_train.columns].copy()
X_submit = bool_encode(X_submit, bool_cols)
X_submit[categorical_cols] = category_encoder.transform(X_submit[categorical_cols])

fraud_class_index = list(model.classes_).index(1)
test_proba = model.predict_proba(X_submit)[:, fraud_class_index]
probabilities_by_id = pd.Series(test_proba, index=test_df['id'])
assert test_df['id'].is_unique and submission['id'].is_unique
assert len(test_df) == len(submission)
submission['is_fraud'] = submission['id'].map(probabilities_by_id)
assert submission['is_fraud'].notna().all()
assert submission['is_fraud'].between(0, 1).all()

submission.to_csv('submission.csv', index=False)
print(f'Da luu {len(submission):,} dong vao submission.csv')
submission.head()
```
